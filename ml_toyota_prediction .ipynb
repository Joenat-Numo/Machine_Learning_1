## Preprocessing of Our Data
- Remove the 'Unamed 0' 

```python
toyota_df.columns
```

```python
toyota_df = toyota_df.drop(columns=['Unnamed: 0'], axis=1)
```

```python
toyota_df
```

```python
## lets convert the KM column to numeric 
toyota_df['KM'] = pd.to_numeric(toyota_df['KM'], errors='coerce')
```

```python
toyota_df.dtypes
```

```python
## lets convert the HP column to numeric 
toyota_df['HP'] = pd.to_numeric(toyota_df['HP'], errors='coerce')
```

```python
toyota_df.dtypes
```

```python
## lets convert the Doors column to numeric 
```

```python
toyota_df
```

```python
## lets replace the value three with 3, four with 4, and five with 5. 
toyota_df['Doors'] = toyota_df['Doors'].replace('three', 3)
toyota_df['Doors'] = toyota_df['Doors'].replace('four', 4)
toyota_df['Doors'] = toyota_df['Doors'].replace('five', 5)
```

```python
## lets convert the HP column to numeric 
toyota_df['Doors'] = pd.to_numeric(toyota_df['Doors'], errors='coerce')
```

```python
toyota_df.dtypes
```

```python
toyota_df.isnull().sum()
```

### Replacing Missing Values 
- Numeric : Mean, Mode, Midean
- Categorical: Group each Cat, Highest Count (Mode)
    - Red = 10, Green = 6, Black = 3 

```python
## cal for average age
mean_age = toyota_df['Age'].mean() 
```

```python
## replace nan with mean_age
toyota_df['Age'] = toyota_df['Age'].fillna(mean_age) 
```

```python
## replacing missing value for KM
toyota_df['KM'] = toyota_df['KM'].fillna(toyota_df['KM'].mean())
```

```python
## replace missing value for FuelType
toyota_df['FuelType'].value_counts()
```

```python
higest_count_fueltype = toyota_df['FuelType'].value_counts().index[0]
higest_count_fueltype
```

```python
toyota_df['FuelType'] = toyota_df['FuelType'].fillna(higest_count_fueltype) 
```

```python
toyota_df.isnull().sum()
```

```python
## replace missing value for HP 
## convert the col to category 
toyota_df['MetColor'] = toyota_df['MetColor'].fillna(toyota_df['MetColor'].value_counts().index[0])
```

```python
##toyota_df['MetColor'].mode()
```

```python
##toyota_df['MetColor'].mean()
```

```python
## replace the missing value with average
toyota_df['HP'] = toyota_df['HP'].fillna(toyota_df['HP'].mean())
```

```python
toyota_df.isnull().sum()
```

```python
## lets convert the MetColor to a category type 
toyota_df['MetColor'] = toyota_df['MetColor'].astype('category')
```

```python
toyota_df.dtypes
```

```python
## lets also convert Automatic , FuelType, and Doors all to category...
toyota_df['Automatic'] = toyota_df['Automatic'].astype('category')
toyota_df['Doors'] = toyota_df['Doors'].astype('category')
toyota_df['FuelType'] = toyota_df['FuelType'].astype('category')
```

```python
toyota_df.dtypes
```

## Validation Framework : Splitting Our Data
- Training Dataset: 60% 
- Validation Dataset: 20% 
- Testing Dataset: 20% 

```python
df_train_full, df_test = train_test_split(toyota_df, test_size=0.2, random_state=11)

df_train, df_valid = train_test_split(df_train_full, test_size=0.25, random_state=11)

print(f'Training Dataset: {len(df_train)}')
print(f'Validation Dataset: {len(df_valid)}')
print(f'Testing Dataset: {len(df_test)}')
```

```python
df_train.head()
```

```python
df_test.head()
```

```python
df_valid.head()
```

```python
## select all the target variables 
y_df_train= df_train['Price']
y_df_valid = df_valid['Price']
y_df_test = df_test['Price']
```

```python
## delete the price columns 
del df_train['Price']
del df_valid['Price']
del df_test['Price']
```

```python
df_train.head()
```

### Selection of baseline Features 
- Age, KM, HP, MetColor, Automatic, CC, Doors Weigth

```python
baseline_ft = ['Age', 'KM', 'HP', 'MetColor', 'Automatic', 'CC', 'Doors', 'Weight']
```

```python
## convert all the target columns to a vector 

y_train = y_df_train.values 
y_valid = y_df_valid.values 
y_test = y_df_test.values
```

```python
y_df_train 
```

```python
y_train 
```

```python
df_train[baseline_ft]
```

```python
## lets also convert the features into matrix format 
X_train_bl = df_train[baseline_ft].values 
X_valid_bl = df_valid[baseline_ft].values 
X_test_bl = df_test[baseline_ft].values
```

```python
df_train 
```

```python
X_train_bl
```

## Training ML Models : Regression Models 

```python
## fit the data to the algorithm 

## instantiate the algorithm 
lrm_model = LinearRegression()

## fit the data to the model
lrm_model.fit(X_train_bl, y_train)
```

## Evaluating the Performance of Model 

```python
## generate the training 
y_train_pred = lrm_model.predict(X_train_bl)

## generate the validation prediction 
y_valid_pred = lrm_model.predict(X_valid_bl)
```

```python
## measure the training performance 
train_score = r2_score(y_train, y_train_pred)

print(f"Training Peformance: {train_score * 100}")
```

```python
## measure the validation peformance 
valid_score = r2_score(y_valid, y_valid_pred)

print(f"Validation Peformance: {valid_score * 100}")
```

## Training and Testing of Best Performing Model 

```python
## selecting the target for df_train_full  

y_train_full = df_train_full['Price']
```

```python
## delete the target from the df_train_full 
del df_train_full['Price']
```

```python
## select only the input features for training 
df_train_full_bl = df_train_full[baseline_ft]
df_train_full_bl
```

```python
## generate the test prediction 
y_test_pred = lrm_model_final.predict(X_test_bl)
```

```python
## 
final_score = r2_score(y_test, y_test_pred)

print(f"Final Test Score: {round(final_score * 100, 2)}")
```
